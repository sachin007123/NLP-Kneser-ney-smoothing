```python
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.cluster import AgglomerativeClustering
from sklearn.metrics.pairwise import cosine_similarity
import numpy as np

print("Libraries imported successfully!")
```

```python
# Text documents
documents = [
    "Football is a popular sport",
    "I love playing football",
    "Cricket is my favorite sport",
    "I enjoy playing cricket",
    "Python is a programming language",
    "Java is used for programming",
    "Machine learning is a branch of AI",
    "Artificial intelligence is growing rapidly"
]

print("Documents created successfully!")
print("Number of documents:", len(documents))
```

```python
# Convert text into TF-IDF vectors
vectorizer = TfidfVectorizer()

X = vectorizer.fit_transform(documents).toarray()

print("TF-IDF conversion completed!")
print("Shape of TF-IDF matrix:", X.shape)
```

```python
# ------------------------------------------------
# 1. AGGLOMERATIVE HIERARCHICAL CLUSTERING
# ------------------------------------------------

agg = AgglomerativeClustering(
    n_clusters=3,
    linkage="ward"
)

agg_labels = agg.fit_predict(X)

print("AGGLOMERATIVE CLUSTERING")
print("-------------------------")

for i in range(3):
    print("\nCluster", i + 1)

    for j in range(len(documents)):
        if agg_labels[j] == i:
            print("-", documents[j])
```

```python
# ------------------------------------------------
# 2. DIVISIVE HIERARCHICAL CLUSTERING
# ------------------------------------------------

def divisive_clustering(X, n_clusters):

    clusters = [list(range(len(X)))]

    while len(clusters) < n_clusters:

        # Select the largest cluster
        cluster = max(clusters, key=len)
        clusters.remove(cluster)

        # Calculate cluster center
        center = np.mean(X[cluster], axis=0)

        # Find document farthest from center
        distances = []

        for i in cluster:

            distance = 1 - cosine_similarity(
                [X[i]], [center]
            )[0][0]

            distances.append(distance)

        seed = cluster[np.argmax(distances)]

        # Create two groups
        group1 = [seed]
        group2 = []

        for i in cluster:

            if i == seed:
                continue

            d1 = 1 - cosine_similarity(
                [X[i]], [X[seed]]
            )[0][0]

            d2 = 1 - cosine_similarity(
                [X[i]], [center]
            )[0][0]

            if d1 < d2:
                group1.append(i)
            else:
                group2.append(i)

        clusters.append(group1)

        if group2:
            clusters.append(group2)

    return clusters

print("Divisive clustering function created successfully!")
```

```python
# Perform divisive clustering
div_clusters = divisive_clustering(X, 3)

print("\nDIVISIVE CLUSTERING")
print("-------------------")

for i, cluster in enumerate(div_clusters):

    print("\nCluster", i + 1)

    for index in cluster:
        print("-", documents[index])
```

```python
# ------------------------------------------------
# COMPARISON
# ------------------------------------------------

print("HIERARCHICAL CLUSTERING COMPARISON")
print("----------------------------------")

print("\nAgglomerative Clustering:")
print("Starts with individual documents and merges similar clusters.")

print("\nDivisive Clustering:")
print("Starts with all documents in one cluster and splits them into smaller clusters.")

print("\nBoth methods divide the documents into 3 groups.")
```
